```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# List of categorical columns to explore
categorical_columns = ["student_id", "course_id", "chapter_order"]

# Loop through each categorical column and plot distribution
for col in categorical_columns:
    plt.figure(figsize=(12, 5))

    # Plot using seaborn countplot
    sns.countplot(y=df[col], order=df[col].value_counts().index[:10], palette="viridis")
    plt.xlabel("Count")
    plt.ylabel(col)
    plt.title(f"Distribution of {col} (Top 10)")

    # Show the plot
    plt.show()
```

```python
def boxplot_outliers(df: pd.DataFrame, column: str):
    plt.figure(figsize=(8, 4))
    sns.boxplot(x=df[column])
    plt.title(f"Boxplot for {completion_status}")
    plt.xlabel(column)
    plt.tight_layout()
    plt.show()
```

```python
def histogram_distribution(df: pd.DataFrame, column: str):
    plt.figure(figsize=(8, 4))
    plt.hist(df[column], bins=30)
    plt.title(f"Histogram of {column}")
    plt.xlabel(column)
    plt.ylabel("Frequency")
    plt.tight_layout()
    plt.show()
```
